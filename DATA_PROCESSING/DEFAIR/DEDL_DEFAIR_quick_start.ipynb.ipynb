## Loading data

DEFAIR comes with several built-in data readers that support lazy loading, enabling efficient handling of large datasets.

```python
from defair_data.readers import list_readers
# See available readers
print(list_readers())
```

Use **Dataset.from_source()** to load data from local or remote storage. DEFAIR automatically detects the appropriate reader based on file format.

### Reader Auto-Detection (local)

```python
from defair_data.core import Dataset

# Local file - reader auto-detected from file extension/content
dataset = Dataset.from_source("./input_data/MSG3-SEVI-MSG15-0100-NA-20230806081241.839000000Z-NA.nat")

# Access the underlying xarray Dataset
print(f"Variables: {list(dataset.data.data_vars)}")
#print(f"Dimensions rtd: {dict(dataset.data.dims)}")
print(f"Dimensions xr: {dict(dataset.sizes)}")
```

### Explicit Reader Selection

```python
from defair_data.readers import list_readers

# See available readers
print(list_readers())  # ['msg15nat', 'era5grib', 'mtg_fci_l1c_nc', ...]

# Specify reader explicitly
dataset = Dataset.from_source(
    "./input_data/MSG3-SEVI-MSG15-0100-NA-20230806081241.839000000Z-NA.nat",
    reader="msg15nat",
    #use_dask=True,  # Enable lazy loading with Dask
)
print(f"Variables: {list(dataset.data.data_vars)}")
```
